### Order Size vs Profitability — Key Finding

Profit margin increased consistently with order size. Small orders generated a negative
margin of -1.21%, while medium, large, and very large orders achieved margins of 3.51%,
6.05%, and 7.31%, respectively.

This suggests that larger orders are associated with stronger profitability rather than
lower profitability. However, this analysis shows an association and does not establish
that increasing order size directly causes higher margins.
